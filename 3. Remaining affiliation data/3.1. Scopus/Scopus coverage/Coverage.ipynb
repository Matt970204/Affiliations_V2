```python
import pandas as pd
All_Data = pd.read_excel('/Users/MattJohnson/Desktop/Affiliations/5. Final consolidation of all data/5.A. Data created in 5./5.2./Final affiliations.xlsx') 
QJE_scopus = pd.read_excel('/Users/MattJohnson/Desktop/Affiliations/1. Input Data/Scopus files/QJE_SCOPUS.xlsx')
AER_scopus = pd.read_excel('/Users/MattJohnson/Desktop/Affiliations/1. Input Data/Scopus files/AER_SCOPUS.xlsx')
RES_scopus = pd.read_excel('/Users/MattJohnson/Desktop/Affiliations/1. Input Data/Scopus files/RES_SCOPUS.xlsx')
JPE_scopus = pd.read_excel('/Users/MattJohnson/Desktop/Affiliations/1. Input Data/Scopus files/JPE_SCOPUS.xlsx')
ECTA_scopus = pd.read_excel('/Users/MattJohnson/Desktop/Affiliations/1. Input Data/Scopus files/ECTA_SCOPUS.xlsx')
```

```python
print(All_Data.shape)
print(QJE_scopus.shape)
print(AER_scopus.shape)
print(RES_scopus.shape)
print(JPE_scopus.shape)
print(ECTA_scopus.shape)
```

```python
for i in range(len(QJE_scopus)):
    QJE_scopus.loc[i,['Article_ID']] = QJE_scopus.loc[i,['DOI']].to_string().rsplit('/',1)[-1]
QJE_scopus = QJE_scopus[['Article_ID' ,'affiliations','year']]

for i in range(len(AER_scopus)):
    AER_scopus.loc[i,['Article_ID']] = AER_scopus.loc[i,['DOI']].to_string().rsplit('/',1)[-1]
AER_scopus = AER_scopus[['Article_ID' ,'affiliations','year']]

for i in range(len(RES_scopus)):
    RES_scopus.loc[i,['Article_ID']] = RES_scopus.loc[i,['DOI']].to_string().rsplit('/',1)[-1]
RES_scopus = RES_scopus[['Article_ID' ,'affiliations','year']]

for i in range(len(JPE_scopus)):
    JPE_scopus.loc[i,['Article_ID']] = JPE_scopus.loc[i,['DOI']].to_string().rsplit('/',1)[-1]
JPE_scopus = JPE_scopus[['Article_ID' ,'affiliations','year']]

for i in range(len(ECTA_scopus)):
    ECTA_scopus.loc[i,['Article_ID']] = ECTA_scopus.loc[i,['DOI']].to_string().rsplit('/',1)[-1]
ECTA_scopus = ECTA_scopus[['Article_ID' ,'affiliations','year']]
```

```python
print(QJE_scopus['year'].min())
print(AER_scopus['year'].min())
print(RES_scopus['year'].min())
print(JPE_scopus['year'].min())
print(ECTA_scopus['year'].min())

print(len(QJE_scopus[(QJE_scopus['affiliations'].notnull()) & (QJE_scopus['year'] <2011)]))
print(len(AER_scopus[(AER_scopus['affiliations'].notnull()) & (AER_scopus['year'] <2011)]))
print(len(RES_scopus[(RES_scopus['affiliations'].notnull()) & (RES_scopus['year'] <2011)]))
print(len(JPE_scopus[(JPE_scopus['affiliations'].notnull()) & (JPE_scopus['year'] <2011)]))
print(len(ECTA_scopus[(ECTA_scopus['affiliations'].notnull()) & (ECTA_scopus['year'] <2011)]))
```

```python
scopus_merged = pd.concat([QJE_scopus,AER_scopus,RES_scopus,JPE_scopus,ECTA_scopus], axis=0)
print(scopus_merged.shape)
print(scopus_merged.head())
```

```python
# Merging the masterfile onto the data in order to get the year
scopus_merged['Article_ID'] = scopus_merged['Article_ID'].astype('str')
All_Data['Article_ID'] = All_Data['Article_ID'].astype('str')
data = All_Data.merge(scopus_merged, how = 'left', on = 'Article_ID')
print(data.shape)
```

```python
data.to_excel('/Users/MattJohnson/Desktop/Affiliations/3. Remaining affiliation data/3.1. Scopus/Scopus coveragescopus_coverage.xlsx') 
```
